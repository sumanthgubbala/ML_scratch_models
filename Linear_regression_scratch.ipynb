## Gradient calculation

### single input

```python
import numpy as np

x = 10
y = 30


w = 2
b = 5

pred = x * w + b

error = pred - y

print(error)
```

```python
dw = 2 * ((error) * x)

print(dw)

db = 2 * error

print(db)
```

```python
lr = 0.001
w_old = w

w_new = w_old - (lr * dw)

print(w_new)

b_new = b - (lr* db)

print(b_new)

```

```python
w = 2
b = 5
for epoch in range(1000):
    
    pred = x * w + b
    
    error = pred - y
    
    dw = 2 * (error * x)
    db = 2 *(error)
    
    lr = 0.001
    w = w - (lr * dw)
    b = b - (lr * db)
    
    print(epoch ," -> ", error, " -> ",pred)
    
    # if error > -1 and error < 0 :
    #     break
    
    
```

```python
pred = x * w_new + b_new

print(pred)

error = pred - y

print(error)
```

# Linear Regression Model

```python
d = np.zeros(2)
d
```

```python
class LinearRegression:
    
    def __init__(self):
        self.w = 0.0
        self.b = 0.0
        self.learning_rate = 0.001
    
    def fit(self,x,y):
        
        for epochs in range(10000):
            pred = self.predict(x)
            
            error = pred - y
            
            dw = 2 / len(x) * np.sum(error * x)
            db = 2/len(x) * np.sum(error)
            
            self.w = self.w - (self.learning_rate * dw)
            self.b = self.b - (self.learning_rate * db)
        
    def predict(self,x):
        
        prediction = x *self.w + self.b
        return prediction
```

```python

x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

lr_model = LinearRegression()

lr_model.fit(x,y)

pred = lr_model.predict(x)
```

```python
print(lr_model.w, lr_model.b)
```

```python
from sklearn.metrics import r2_score

print(r2_score(pred, y))
```
